```python
eng=input("Enter english marks: ")
math=input("Enter maths marks: ")
social=input("Enter social marks: ")
sci=input("Enter science marks: ")
french=input("Enter french marks: ")
total=(float(eng)+float(math)+float(social)+float(sci)+float(french))/5
print(total)
if total>=90 and total<=100:
  print("Distinction")
elif total>80 and total<90:
  print("outstanding")
elif total>70 and total<80:
  print("first class")
elif total>60 and total<70:
  print("second class")
elif total>50 and total<60:
  print("third class")
elif total<50:
  print("fail")
else:
  print("error")
```

#OR Operator

```python
number1= int(input("Enter your number1: "))
number2= int(input("Enter your number2:"))
choice = input("Enter your Choice : ")
if choice == "add" or choice == "Add":
  print(number1+number2)
elif choice == "sub" or choice == "Sub":
  print(number1-number2)
elif choice == "mul" or choice == "Mul":
  print(number1*number2)
elif choice == "div" or choice == "Div":
  print(number1/number2)
else:
  print("Invalid Choice")
```

#Pattern Programs

```python
n=int(input("enter the number of rows: "))
for i in range(1,n+1):
  for j in range(1,i+1):
    print("* ", end = " ") #commenting
  print(" ")
```
